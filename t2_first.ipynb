```python
s=input("Enter string: ")
s=s.title()
for i,char in enumerate(s):
    if char==' ':
        s=s[0:i-1]+s[i-1].upper()+s[i:len(s)-1]+s[len(s)-1].upper()
        break
print(s)
```

```python
s=input("Enter sentance: ")
s=s.title()[:-1]+s[-1].upper()
for i,j in enumerate(s):
    if j==" ":
        count=i-1
        break
s=s[0:count]+s[count].upper()+s[count+1:]
print(s)
```
