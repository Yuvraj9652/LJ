## Step 1: Import Required Libraries

Pandas is used for data handling. Seaborn and Matplotlib are used for static plots. Plotly Express is used for interactive plots.

````python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
from io import StringIO

sns.set_theme(style="whitegrid")
pd.set_option("display.max_columns", None)
````

## Step 2: Read the Embedded CSV Dataset

The CSV data is stored as text inside the notebook and read using `StringIO`.

````python
sales_csv = """OrderID,Month,Region,Category,Sales,Profit,Discount,Quantity,Rating,DeliveryDays
1001,Jan,West,Electronics,72000,8500,0.10,1,4.6,3
1002,Jan,North,Furniture,18000,2400,0.00,3,4.1,5
1003,Jan,South,Electronics,12500,1800,0.05,5,4.4,4
1004,Jan,East,Clothing,9600,1100,0.15,4,3.8,6
1005,Feb,West,Grocery,8500,900,0.02,10,3.9,2
1006,Feb,North,Electronics,68000,7600,0.08,2,4.8,3
1007,Feb,South,Furniture,30000,4200,0.00,2,4.0,7
1008,Feb,East,Clothing,15000,2100,0.12,6,4.3,4
1009,Mar,West,Electronics,54000,6900,0.07,3,4.7,3
1010,Mar,North,Grocery,14400,1600,0.03,8,3.8,5
1011,Mar,South,Clothing,7200,950,0.20,12,4.2,6
1012,Mar,East,Electronics,48000,6100,0.09,4,4.5,4
1013,Apr,West,Furniture,22000,3300,0.06,2,3.7,8
1014,Apr,North,Clothing,13300,1750,0.00,7,4.0,5
1015,Apr,South,Grocery,11700,1400,0.04,9,4.1,3
1016,Apr,East,Furniture,42000,5200,0.11,1,4.1,7
1017,May,West,Electronics,56000,7200,0.10,2,4.9,2
1018,May,North,Grocery,15600,1900,0.05,6,4.4,4
1019,May,South,Furniture,38000,4600,0.08,1,3.9,8
1020,May,East,Clothing,16000,2300,0.18,10,4.6,5
1021,Jun,North,Electronics,36000,5100,0.06,4,4.5,3
1022,Jun,West,Furniture,28000,3900,0.07,1,4.2,6
1023,Jun,South,Clothing,18400,2600,0.10,8,4.3,4
1024,Jun,East,Grocery,13200,1500,0.03,11,4.0,3
"""

sales = pd.read_csv(StringIO(sales_csv))
sales.head()
````

## Step 3: Basic Dataset Information

Check the number of rows and columns, data types, missing values, and numeric summary.

````python
print("Dataset shape:", sales.shape)
print("\nDataset information:")
sales.info()

print("\nSummary statistics:")
display(sales.describe())
````

## Step 4: Seaborn Box Plot

This box plot compares the distribution of sales across product categories. It helps identify which categories have higher sales spread and possible outliers.

````python
plt.figure(figsize=(9, 5))
sns.boxplot(data=sales, x="Category", y="Sales", hue="Category", palette="Set2", legend=False)
plt.title("Sales Distribution by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Sales")
plt.tight_layout()
plt.show()
````

## Step 5: Seaborn Scatter Plot

This scatter plot shows the relationship between sales and profit. Region is shown by color, and category is shown by marker style.

````python
plt.figure(figsize=(9, 6))
sns.scatterplot(
    data=sales,
    x="Sales",
    y="Profit",
    hue="Region",
    style="Category",
    s=110
)
plt.title("Sales vs Profit by Region and Category")
plt.xlabel("Sales")
plt.ylabel("Profit")
plt.tight_layout()
plt.show()
````

## Step 6: Seaborn Heatmap

The heatmap shows correlation among numeric columns. Values close to `1` mean strong positive relation, values close to `-1` mean strong negative relation, and values near `0` mean weak relation.

````python
numeric_data = sales.select_dtypes(include="number")
correlation_matrix = numeric_data.corr()

plt.figure(figsize=(9, 6))
sns.heatmap(correlation_matrix, annot=True, cmap="coolwarm", fmt=".2f", linewidths=0.5)
plt.title("Correlation Heatmap of Numeric Columns")
plt.tight_layout()
plt.show()
````

## Step 7: Plotly Bar Chart

This interactive bar chart shows total sales by region.

````python
region_sales = sales.groupby("Region", as_index=False)["Sales"].sum()

fig_bar = px.bar(
    region_sales,
    x="Region",
    y="Sales",
    color="Region",
    title="Total Sales by Region",
    text="Sales"
)
fig_bar.update_traces(textposition="outside")
fig_bar.show()
````

## Step 8: Plotly Line Chart

This interactive line chart shows how total sales changed month by month.

````python
month_order = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
monthly_sales = sales.groupby("Month", as_index=False)["Sales"].sum()
monthly_sales["Month"] = pd.Categorical(monthly_sales["Month"], categories=month_order, ordered=True)
monthly_sales = monthly_sales.sort_values("Month")

fig_line = px.line(
    monthly_sales,
    x="Month",
    y="Sales",
    markers=True,
    title="Monthly Sales Trend"
)
fig_line.show()
````

## Step 9: Plotly Scatter Plot

This interactive scatter plot compares discount and profit. Larger points indicate higher sales, and color represents product category.

````python
fig_scatter = px.scatter(
    sales,
    x="Discount",
    y="Profit",
    size="Sales",
    color="Category",
    hover_data=["OrderID", "Month", "Region", "Sales", "Quantity"],
    title="Discount vs Profit by Category"
)
fig_scatter.show()
````

## Step 10: Plotly Histogram

This interactive histogram shows the distribution of customer ratings, separated by region.

````python
fig_hist = px.histogram(
    sales,
    x="Rating",
    color="Region",
    nbins=8,
    barmode="overlay",
    title="Customer Rating Distribution by Region"
)
fig_hist.update_traces(opacity=0.75)
fig_hist.show()
````

## Step 11: Conclusion

From the visualizations:

- Electronics generally has higher sales values compared with other categories.
- Sales and profit show a clear positive relationship: higher sales usually produce higher profit.
- The heatmap confirms that `Sales` and `Profit` are strongly correlated.
- Regional sales can be compared clearly using the Plotly bar chart.
- Monthly sales trend helps identify sales growth or decline over time.
- Discount and profit scatter analysis helps check whether higher discounts reduce profit.
- Rating distribution helps compare customer satisfaction across regions.
