```python
city_name = "Ahmedabad"
API_KEY = "740de831eb297883868807ace513fb2a"
BASE_URL = f"http://api.openweathermap.org/geo/1.0/direct?q={city_name}&appid={API_KEY}"
response = requests.get(BASE_URL)
if response.status_code == 200:
    data = response.json()
    print(data)
else:    
    print("Error in the HTTP request",response.status_code)
```

```python
# for item in data:
#     print("Latitude:",item['lat'])
#     print("Longitude:",item['lon'])
print("City Name:",data[0]['name'])
print("Latitude:",data[0]['lat'])
print("Longitude:",data[0]['lon'])
lat = data[0]['lat']
lon = data[0]['lon']
```

```python
WEATHER_API=f"https://api.openweathermap.org/data/2.5/weather?lat={lat}&lon={lon}&appid={API_KEY}"
response = requests.get(WEATHER_API)
data = response.json()
print(data)
```
