```python
box=soup.find('article',class_="main-article")
print(len(box.find_all("a")))
```

```python
links=[]
for i in box.find_all("a"):
    links.append("https://subslikescript.com" + i["href"])
print(links)
```

```python
import os
for i in links:
    response=requests.get(i)
    soup=BeautifulSoup(response.text,'html.parser')
    title=soup.find('article',class_="main-article").find('h1').text
    script=soup.find('div',class_="full-script").text.strip()
    # Store all scripts in a folder named "scripts" after making folder named "scripts" in the current directory
    os.makedirs("scripts", exist_ok=True)
    with open(f"scripts/{title}.txt","w",encoding="utf-8") as f:
        f.write(script)
    print(title)
```

```python
for i in range(1,4):
    url=f"https://subslikescript.com/movies_letter-V?page={i}"
    response=requests.get(url)
    soup=BeautifulSoup(response.text,'html.parser')
    box=soup.find('article',class_="main-article")
    links=[]
    for i in box.find_all("a"):
        links.append("https://subslikescript.com" + i["href"])
    for i in links:
        response=requests.get(i)
        soup=BeautifulSoup(response.text,'html.parser')
        title=soup.find('article',class_="main-article").find('h1').text
        script=soup.find('div',class_="full-script").text.strip()
        # Store all scripts in a folder named "scripts" after making folder named "scripts" in the current directory
        os.makedirs("scripts", exist_ok=True)
        page_num = url.rsplit('=', 1)[-1]
        os.makedirs(f"scripts/page_{page_num}", exist_ok=True)
        # Use try-except block to handle any errors that may occur while writing the file and continue with the next one
        try:
            with open(f"scripts/page_{page_num}/{title}.txt","w",encoding="utf-8") as f:
                f.write(script)
        except Exception as e:
            print(f"Error occurred while writing {title}: {e}")
        print(title)
```

```python
url="https://webscraper.io/test-sites/e-commerce/allinone/computers/tablets"
response=requests.get(url)
soup=BeautifulSoup(response.text,'html.parser')
```

```python
# create csv coloums product_name, price, description
products=soup.find_all("div",class_="caption")
data=[]
for product in products:
    name=product.find("a",class_="title").text.strip()
    price=product.find("h4",class_="pull-right").text.strip()
    description=product.find("p",class_="description card-text").text.strip()
    data.append({"product_name":name,"price":price,"description":description})
print(data)
df=pd.DataFrame(data)
```

```python
df.to_csv("products.csv",index=False)
```
