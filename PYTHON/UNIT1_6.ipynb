```python
Q1=auto.mpg.quantile(0.25)
Q1
```

```python
Q2=auto.mpg.quantile(0.5)
Q2
```

```python
Q3=auto.mpg.quantile(0.75)
Q3
```

```python
auto.mpg.describe()
```

```python
IQR=Q3-Q1
IQR
```

```python
HL=Q3+1.5*IQR
LL=Q1-1.5*IQR
```

```python
HL,LL
```

```python
auto.mpg.describe()
```

```python
wo=auto[(auto["mpg"]<HL) & (auto["mpg"]>LL)]
wo
```

```python
sns.boxplot(wo["mpg"])
```

```python
sns.boxenplot(wo["mpg"])
```
