```python
import pandas as pd
```

```python
df=pd.read_csv("diabetes.csv")
df.head()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, accuracy_score
from sklearn.tree import DecisionTreeClassifier,plot_tree
```

```python
x=df.drop("Outcome",axis=1)
y=df["Outcome"]
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
for i in range(1,50):
    dt=DecisionTreeClassifier(criterion="entropy",random_state=42,max_depth=i)
    dt.fit(x_train,y_train)
    y_pred=dt.predict(x_test)
    print(accuracy_score(y_test,y_pred))
    # print(classification_report(y_test,y_pred))
    # plot_tree(dt,feature_names=x.columns,filled=True)
```

```python
import matplotlib.pyplot as plt
from sklearn.neighbors import KNeighborsClassifier
k=31
accuracy=[]
test_sizes=[0.05,0.1,0.15,0.2,0.25]
x=df.drop("Outcome",axis=1)
y=df["Outcome"]
for i in test_sizes:
    x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=i,random_state=42)
    dt=DecisionTreeClassifier(criterion="entropy",random_state=42,max_depth=5)
    dt.fit(x_train,y_train)
    y_pred=dt.predict(x_test)
    acc=accuracy_score(y_test,y_pred)
    accuracy.append(acc)
    print(acc)
plt.plot(test_sizes,accuracy)
plt.xlabel("Test Size")
plt.ylabel("Accuracy")
plt.show()
```

```python
import matplotlib.pyplot as plt
from sklearn.neighbors import KNeighborsClassifier
k=31
accuracy=[]
test_sizes=[0.05,0.1,0.15,0.2,0.25]
x=df.drop("Outcome",axis=1)
y=df["Outcome"]
for i in test_sizes:
    x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=i,random_state=42)
    knn=KNeighborsClassifier(n_neighbors=k)
    knn.fit(x_train,y_train)
    y_pred=knn.predict(x_test)
    acc=accuracy_score(y_test,y_pred)
    accuracy.append(acc)
    print(acc)
plt.plot(test_sizes,accuracy)
plt.xlabel("Test Size")
plt.ylabel("Accuracy")
plt.show()
```
