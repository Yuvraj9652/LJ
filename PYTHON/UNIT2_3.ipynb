```python
df["Total"]=df.apply(sum,axis=1)
df
```

```python
df["Grade"]=df.apply(lambda x:"A" if x["Total"]>300 else ("B" if x["Total"]>200 else "C"),axis=1)
df
```

```python
def grade(x):
    if x>300:
        return "A+"
    if x>200:
        return "A"
    if x>100:
        return "B"
    return "F"
df["GradeUsin.apply"]=df["Total"].apply(grade);
df
```

```python
df["Indiviadual"]=df.apply(lambda x:"Fail" if ((x["A"]<35) or (x["B"]<35) or (x["C"]<35) or (x["D"]<35) or (x["E"]<35)) else "Pass",axis=1)
df
```

```python
def res(x):
    if (x["A"]<35) or (x["B"]<35) or (x["C"]<35) or (x["D"]<35) or (x["E"]<35):
        return "Fail"
    return "Pass"
df.apply(res,axis=1)
```

```python
df["Percentage"]=df.apply(lambda x:x["Total"]*0.2,axis=1)
df
```

```python
df["Result"]=df.apply(lambda x:"Fail" if x["Percentage"]<35.0 else "Pass",axis=1)
df
```
