```python
import pandas as pd
import numpy as np
import seaborn as sns
```

```python
df=pd.read_csv("PANDAS_DATA/social_media_engagement.csv")
df
```

```python
df.corr(numeric_only=True)
```

```python
sns.heatmap(df.corr(numeric_only=True))
```

```python
sns.heatmap(df.corr(numeric_only=True),cmap="coolwarm")
```

```python
sns.heatmap(df.corr(numeric_only=True),cmap="coolwarm",linecolor="red",linewidths=5)
```

```python
sns.heatmap(df.corr(numeric_only=True),cmap="coolwarm",linecolor="red",linewidths=5,annot=True)
```

```python
sns.heatmap(df.corr(numeric_only=True),cmap="coolwarm",linecolor="red",linewidths=5,cbar=False)
```

```python
sns.heatmap(df.corr(numeric_only=True),cmap="coolwarm",linecolor="red",linewidths=5,annot=True,vmax=0.8,vmin=0.3)
#makes same color for specific range using vmax,vmin
```
