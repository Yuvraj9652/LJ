```python
class Atm:
    def __init__(self):
        self.menu()
    def menu(self):
        while True:
            print('''
            1.)Enter Pin
            2.)Change Pin
            3.)Withdraw
            4.)Exit''')
            n=int(input("Enter choice: "))
            if n==1:
                self.createpin()
            elif n==2:
                self.changepin()
            elif n==3:
                self.withdraw()
            elif n==4:
                break
    def createpin(self):
        self.pin=self.enterpin()
        self.balance=float(input("Enter balance: "))
    def withdraw(self):
        self.withdraw=float(input("Enter amount to witdraw: "))
        if self.withdraw<=self.balance:
            self.balance-=self.withdraw
            print("New Balance: ",self.balance)
        else:
            print("Insufficent amount")
    def enterpin(self):
        pin=input("Enter pin: ")
        return pin
    def checkpin(self):
        return enterpin()==self.pin
    def changepin(self):
        if self.checkpin():
            self.pin=enterpin()
        else:
            print("Wrong pin")
            return
a=Atm()

##Here you would learn that you should always define variables in constructor
```
