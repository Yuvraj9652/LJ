```python
balance=10000
try:
    withd=int(input("Enter amount to withdraw: "))
    if withd<0:
        raise Exception("Enter Positive")
    elif balance-withd<0:
        raise Exception("Insufficient Money")    
    balance-=withd
    print(balance," amount left")
except Exception as e:
    print(e)
```

```python
class Vishal(Exception):
    def __init__(self,message):
        pass
balance=1000
try:
    withd=int(input("Enter amount to withdraw: "))
    if withd<0:
        raise Exception("Enter Positive")
    elif balance-withd<0:
        raise Vishal("Insufficient Money")    
    balance-=withd
    print(balance," amount left")
except Exception as e:
    print(e)
```
