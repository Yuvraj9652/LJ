# Scatter Plot

```python
x = [1,2,3,4,5]
y = [10,20,30,40,50]
z = [15,25,35,45,55]
plt.scatter(x,y)
plt.scatter(x,z)
plt.scatter(y,z)
plt.show()
```

```python
x = ['T1','T2','T3','T4']
y = [15,25,35,45]
z = [15,25,30,40]
c1 = ['green','blue','black','yellow']
s1 = [200,200,500,1000]
plt.scatter(x,y,c = c1,s = s1,alpha=0.2)
plt.scatter(x,z,c = c1,alpha=0.9)
plt.show()
```

# BarGraph

```python
x = [1,2,3,4,5]
y = [10,20,30,15,5]
c1 = ['red','blue','green','yellow']
plt.barh(x,y,color=c1,height=0.2)
plt.show()
```

# PieChart

```python
x = [1,2,3,4,5]
label = ['t1','t2','t3','t4','t5']
e = [0.1,0.1,0.1,0.4,0.1]
plt.pie(x,startangle=90,labels=label,explode=e,autopct="%1.f")
plt.legend(title='marks',loc='right')
plt.show()
```

```python
y = [80,80,90,90]
x = ['T1','T2','T3','T4']
f1 = {'family':'serif','color':'green','size':20}
plt.subplot(2,3,1)

plt.plot(x,y,marker='o',ms=1,color='black',linewidth=2,linestyle=':')
plt.title('Analysis',fontdict=f1)
plt.xlabel('Phases',fontdict=f1)
plt.ylabel('Marks',fontdict=f1)

plt.subplot(2,3,2)
s1 = [100,100,100,100]
c1 = ['red','blue','yellow','green']
plt.scatter(x,y,c=c1,s=s1,alpha=0.7)

plt.subplot(2,3,3)
c1 = ['red','blue','green','yellow']
plt.bar(x,y,color=c1,width=0.5)

plt.subplot(2,3,4)
c1 = ['red','blue','green','yellow']
plt.barh(x,y,color=c1,height=0.5)



plt.show()
```
