```python
n=input("Enter n: ")
t= False
if 'done' in n or 'Done' in n:
    t=True
else:
    n1=int(n)
    max=n1
    min=n1
    avg=n1
    sum=n1
    count=1
while True:
    if  t:
        break
    n=input("Enter n: ")
    if 'done' in n or 'Done' in n:
        break
    n1=int(n)
    if n1>max:
        max=n1
    elif n1<min:
        min=n1
    sum+=n1
    count+=1
if t:
    print("Empty")
else:
    avg=sum/count
    print(f"{max} {min} {avg}")
```
