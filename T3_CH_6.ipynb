```python
f=open("logfile.txt")
for i in f:
    x=i.split()
    start=x[1].split(":")
    end=x[2].split(":")
    if (int(end[0])-int(start[0]))>1:
        print(x[0])
    if (int(end[0])-int(start[0]))==1:
        if int(end[1])>=int(start[1]):
            print(x[0])
f.close()
```

```python
f=open("namelist.txt")
inp=input("Enter abr.: ")
for i in f:
    x=i.split()
    if len(x)==2:
        if x[0][0]==inp[0] and x[1][0]==inp[1]:
            print(i)
    if len(x)==3:
        if len(inp)==2:
            if (x[0][0]==inp[0] and x[2][0]==inp[1]):
                print(i)
        else:
            if (x[0][0]==inp[0] and x[1][0]==inp[1] and x[2][0]==inp[2]):
                print(i)
```
