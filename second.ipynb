```python
def reverse_num(n):
    "reverse Int"
    sum=0
    while n!=0:
        sum=sum*10+n%10
        n=n//10
    return sum
n=int(input("Enter n: "))
print(reverse_num(n))
```

```python
def harshad_number_confirmation(n):
    "Harshad number"
    n1=n
    sum=0
    while n1!=0:
        sum=sum+n1%10
        n1=n1//10
    if n%sum==0:
        return True
    else:
        return False
def harshad_number_1_n(n):
    l=[]
    for i in range(1,n+1):
        if harshad_number_confirmation(i):
            l.append(i)
    return l
print(harshad_number_1_n(100))
```
