```python
i=1
max1=0
min1=0
max2=0
min2=0
sum=0
mark=int(input(f"Enter marks of Student {i}(enter -1 to exit): "))
if mark==-1:
    print("No Marks available")
else:
    if mark>100 or mark<0:
        print("Invalid Input")
    else:
        max1=mark
        min1=mark
        sum=mark
        i=2
    while True:
        marks=int(input(f"Enter marks of Student {i}(enter -1 to exit): "))
        if marks==-1:
            break
        if marks>100 or marks<0:
            print("invalid marks")
            continue
        if i==1:
            if max1>marks:
                max2=marks
                min2=min1
                min1=marks
            else:
                max2=max1
                max1=marks
                min2=marks
        else:
            if max2<marks:
                if max1<marks:
                    max2=max1
                    max1=marks
                else:
                    max2=marks
            if min2>marks:
                if min1>marks:
                    min2=min1
                    min1=marks
                else:
                    min2=marks
        i+=1
        sum+=marks
    avg=sum/i
    if max2!=0:
        print(f"Max={max1} Max2={max2} \nMin={min1} Min2={min2} \nAvg={avg}") 
    else:
        print(f"Max={max1} Min={min1} Avg={avg}")
```
