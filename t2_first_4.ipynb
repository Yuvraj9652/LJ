```python
# password check question
    # length>8
    # atleast 2 upper
    # atleast 2 lower
    # atleast 2 digit
    # atleast 2 sp. char
    # space not allowed
s=input("Enter pass: ")
if len(s)<=8:
    print("More than 8 letter")
else:
    a=b=c=0
    x=True
    for i in s:
        if i.isspace():
            print("It must not cotain space")
            x=False
            break
        if i.islower():
            a+=1
        if i.isupper():
            b+=1
        if i.isdigit():
            c+=1
    d=len(s)-a-b-c
    if x:
        if a<2:
            print("It must contain atleast 2 upper alphabets")
        if b<2:
            print("It must contain atleast 2 upper alphabets")
        if c<2:
            print("It must contain atleast 2 digits")
        if d<2:
            print("It must contain atleast 2 speacial char")
        if a>1 and b>1 and c>1 and d>1:
            print("Valid password")
```
