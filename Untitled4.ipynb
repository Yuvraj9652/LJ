```python
class Atm:
    def __init__(self):
        self.pin=""
        self.balance=50000
        print("Automatic")
        print(id(self))
    def menu(self):
        choice=input('''
        1.setpin
        2.changepin
        3.withdraw
        4.deposit
        5.checkbalance
        6.exit'
        Enter choice: ''')
obj1=Atm()
print(obj1.balance)
print(id(obj1))
obj1.menu()
obj2=Atm()
print(id(obj2))
```

```python
class Atm:
    def __init__(self,pin,bal):
        self.pin=pin
        self.bal=bal
    def menu(self):
        choice=input('''
        1.setpin
        2.changepin
        3.withdraw
        4.deposit
        5.checkbalance
        6.exit'
        Enter choice: ''')
o1=Atm(3252,10)
print(o1.pin)
print(o1.bal)
print(id(o1.bal))
o1.bal=100000
print(id(o1.bal))#object is muttable
```
