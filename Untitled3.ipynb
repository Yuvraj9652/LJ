```python
f=open("secretmessage.txt")
f1=open("expo.txt","w+")
for i in f:
    count=0
    for j in i.split():
        count+=len(j)
        if count>80:
            f1.write("\n"+j)
            print(j)
            count=0
        else:
            f1.write(j+" ")
            count+=1
            print(j)
        print(count)
f.close()
f1.close()
```
